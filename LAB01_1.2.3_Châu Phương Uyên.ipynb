# Thực hiện trực quan hóa dữ liệu trên tập dữ liệu về bệnh tiểu đường

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
diabetes_data = pd.read_csv("diabetes.csv")
```

```python
diabetes_data.head()
```

```python
diabetes_data = diabetes_data[['Pregnancies', 'Glucose', 'BloodPressure', 'BMI', 'Age', 'Outcome']]
```

Thư viện Matplotlib - Case 1

```python
outcome_count = diabetes_data['Outcome'].value_counts()
print(outcome_count)
```

```python
x = ['No Diabetes', 'Diabetes']
y = outcome_count.values
print(x)
print(y)
```

```python
plt.figure(figsize=(8,5))
plt.bar(x, y)

plt.show()
```

Thư viện Matplotlib - Case 2

```python
plt.figure(figsize=(8,5))
plt.bar(x, y)
plt.title('Number of Diabetes Cases', fontsize=15)
plt.xlabel('Outcome', fontsize=12)
plt.xticks(fontsize=10)
plt.ylabel('Count', fontsize=12)
plt.yticks(fontsize=10)

plt.show()
```

Thư viện Matplotlib - Case 3

```python
fig, ax = plt.subplots(figsize=(15,6))

plt.subplot(1,2,1)
plt.hist(diabetes_data['Glucose'], bins=20)
plt.title('Distribution of Glucose', fontsize=15)
plt.xlabel('Glucose', fontsize=12)
plt.ylabel('Count', fontsize=12)

plt.subplot(1,2,2)
plt.hist(diabetes_data['BMI'], bins=20)
plt.title('Distribution of BMI', fontsize=15)
plt.xlabel('BMI', fontsize=12)
plt.ylabel('Count', fontsize=12)

plt.show()
```

Thư viện Seaborn - Case 1

```python
plt.figure(figsize=(8,5))
sns.countplot(data=diabetes_data,x='Outcome')

plt.show()
```

Thư viện Seaborn - Case 2

```python
plt.figure(figsize=(8,5))
ax = sns.countplot(data=diabetes_data,x='Outcome')
ax.set_xlabel('Outcome',fontsize=15)
ax.set_ylabel('Count',fontsize=15)
ax.set_title('Number of Diabetes Cases',fontsize=20)

plt.show()
```

Thư viện Seaborn - Case 3

```python
fig, ax = plt.subplots(1,2,figsize=(18,6))
sns.set(font_scale=1.2)
ax1 = sns.histplot(data=diabetes_data,x='Glucose',ax=ax[0])
ax1.set_xlabel('Glucose')
ax1.set_ylabel('Count')
ax1.set_title('Distribution of Glucose')
ax2 = sns.histplot(data=diabetes_data,x='BMI',ax=ax[1])
ax2.set_xlabel('BMI')
ax2.set_ylabel('Count')
ax2.set_title('Distribution of BMI')

plt.show()
```

```python
retail_data = retail_data[['Quantity','UnitPrice','Country']]
```

Thư viện Matplotlib - Case 1

```python
country_sales = retail_data.groupby('Country')['Quantity'].sum().sort_values(ascending=False).head(10)
country_sales
```

```python
x = country_sales.index
y = country_sales.values
```

```python
plt.figure(figsize=(12,6))
plt.bar(x,y)

plt.show()
```

Thư viện Matplotlib - Case 2

```python
plt.figure(figsize=(12,6))
plt.bar(x,y)
plt.title('Top 10 Countries by Quantity Purchased',fontsize=15)
plt.xlabel('Country',fontsize=12)
plt.ylabel('Quantity',fontsize=12)
plt.xticks(rotation=45)

plt.show()
```

Thư viện Matplotlib - Case 3

```python
fig, ax = plt.subplots(figsize=(15,6))
plt.subplot(1,2,1)
plt.hist(retail_data['Quantity'],bins=30)
plt.title('Distribution of Quantity')
plt.subplot(1,2,2)
plt.hist(retail_data['UnitPrice'],bins=30)
plt.title('Distribution of Unit Price')

plt.show()
```

Thư viện Seaborn - Case 1

```python
plt.figure(figsize=(12,6))
country_sales = retail_data.groupby('Country')['Quantity'].sum().sort_values(ascending=False).head(10)
data = country_sales.reset_index()
sns.barplot(data=data,x='Country',y='Quantity')

plt.show()
```

Thư viện Seaborn - Case 2

```python
plt.figure(figsize=(12,6))
data = country_sales.reset_index()
ax = sns.barplot(data=data,x='Country',y='Quantity')
ax.set_xlabel('Country',fontsize=15)
ax.set_ylabel('Quantity',fontsize=15)
ax.set_title('Top 10 Countries by Quantity Purchased',fontsize=20)
plt.xticks(rotation=45)

plt.show()
```

Thư viện Seaborn - Case 3

```python
clean_data = retail_data[
    (retail_data['Quantity'] > 0) &
    (retail_data['Quantity'] <= 100) &
    (retail_data['UnitPrice'] > 0) &
    (retail_data['UnitPrice'] <= 100)
]

fig, ax = plt.subplots(1, 2, figsize=(18,6))
sns.histplot(data=clean_data,x='Quantity',ax=ax[0])
ax[0].set_title('Distribution of Quantity')
sns.histplot(data=clean_data,x='UnitPrice',ax=ax[1])
ax[1].set_title('Distribution of Unit Price')

plt.show()
```
